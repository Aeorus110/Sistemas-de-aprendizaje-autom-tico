```python
# Bloque de importaciones
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Lectura del DataSet añadiendo el nombre de las columnas
df = pd.read_csv('./processed.cleveland.data',
names=['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'num']
)

# Se Presenta información del dataset
print("Filas y columnas:",df.shape)  # Cuantas filas y columnas tiene el dataset
print()
```

```python
print("Tipo de datos:")
print(df.info()) # Tipos de datos de cada columna (característica o atributo)
print()
```

```python
print("Valores para las 20 primeras filas:") 
print(df.head(20).to_string()) # Devuelve las 20 primeras filas sin cortes tabulares (todo en una sola línea)
print()
```

```python
print("Descripción estadística de los datos:")
print(df.describe())
```

```python
# Muestra las filas que tienen un valor no numérico en las columnas ca y thal
df[pd.to_numeric(df.ca, errors='coerce').isna() | pd.to_numeric(df.thal, errors='coerce').isna()]
```

```python
# Modifica el tipo de estos atributos a numérico. Podrían ser números pero en formato texto.
# Convierte en valores NaN todos los valores no numéricos de los atributos numéricos ca y thal
# errors='coerce'	Convierte en NaN los valores que no pueden transformarse a números por ejemplo ?

df.ca = pd.to_numeric(df.ca, errors='coerce') # Guarda el resultado en la misma columna, sustituyendo sus valores anteriores.
df.thal = pd.to_numeric(df.thal, errors='coerce')

# Verificamos que las columnas ca y thal son ahora de un tipo numérico y vemos el número de valores No nulos que contienen.
# ¿Cuántos valores NULOS contienen?
print(df[['ca', 'thal']].isnull().sum())
# Contiene 6 nulos
# Se puede verificar también que estos valores nulos son ahora valores NaN.
df.info()

print()
```

```python
# Muestra las filas que tienen un valor no numérico en las columnas ca y thal ahora corregidos
df[pd.to_numeric(df.ca, errors='coerce').isna() | pd.to_numeric(df.thal, errors='coerce').isna()]
```

```python
# Una vez corregidos los valores nulos, volvemos a ejecutar la función df.describe()
# Pregunta:
#   ¿Qué información útil para el proceso que estamos llevando a cabo (análisis y limpieza de datos) obtenemos?
# Conteo de nulos y sesgos(mean, min, max)
#   ¿Cuántos valores no nulos tiene la columna ca?
# Tiene 299 valores no nulos.
#   ¿Cuántos valores no nulos tiene la columna thal?
# Tiene 301 valores no nulos.
print(df.describe())
```

```python
# Añade a continuación las instrucciones necesarias para eliminar las filas que contienen valores nulos en algún atributo, sin
# modificar el df original. Almacena el nuevo df en df1 y muestra las filas con valores nulos contenidos en df1
# Añade aquí el código......
df1 = df.dropna()
print("Valores nulos: ")
print(df1.isnull().sum())

```

```python
# Añade aquí el código necesario para sustituir los valores nulos en df por la média de la columna edad.
# Añade aquí el código.....
df = df.fillna(df['age'].mean())
print('Nulos: ')
print(df[['ca', 'thal']].isnull().sum())
```

```python
# Covertimos un atributo numérico a categórico
df['num']=pd.Categorical(df['num'])
df.info()
```

```python
# Convertir un tributo categórico a numérico
df['num'] = pd.to_numeric(df['num'], errors='coerce') # con el argumento 'coerce', si el valor no lo puedo convertir genero un NaN
df.info()
```



## Diagramas de barras para atributos categóricos

```python
# Frecuencia absoluta para cada uno de los valores del atributo num 
print(df['num'].value_counts())
```

```python
# Frecuencia relativa
print(100 * df['num'].value_counts()/df.shape[0])
```

```python
# Mostrar un diagrama de barras para estos datos ordenados por el valor del atributo
df['num'].value_counts().sort_index().plot(kind='bar')
```

```python
# Mostrar un diagrama de barras para estos datos ordenados por el valor de frecuencia (en este caso coinciden)
df['num'].value_counts().plot(kind='bar')
```

## Histogramas para atributos numéricos

```python
# Mostramos los histogramas de atributos numéricos
df.hist()
plt.tight_layout() # Con esto auto-regulamos la distribución en pantalla de los histogramas
```

```python
# Mostramos el histograma para el atributo 'age' sin y con especificación de bins. Después el histograma de 4 atributos en particular
df.hist(['age'])
df.hist(['age'],  bins=30)
df.hist(['age','trestbps','chol','oldpeak'],  bins=30)
```

```python
# Histogramas con plt
plt.hist(df['age'])
plt.xlabel('edad') 
plt.ylabel('frec. absoluta') 
plt.xticks(np.arange(0,100,5)) 
plt.gcf().set_size_inches(8, 4)
plt.gca().tick_params(axis='x', labelrotation=45)
```
