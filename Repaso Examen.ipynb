Enunciado: Imagina que tienes un archivo llamado ventas_tienda.csv. El archivo tiene las 3 primeras líneas con información de cabecera que debes ignorar. Además, usa el punto y coma (;) como separador de columnas, y algunas celdas defectuosas contienen el texto "SIN_DATOS" que debe interpretarse como un valor nulo (NaN). Escribe el código en Python para:

Cargar el CSV aplicando correctamente los ajustes mencionados en un DataFrame llamado df.

Mostrar un resumen con el tipo de datos de cada columna y el recuento de valores no nulos.

Eliminar la columna "CodigoTienda" del DataFrame sin necesidad de reasignarlo a una nueva variable (directamente sobre el original).

Mostrar las primeras 7 filas del DataFrame resultante.

```python
import pandas as pd

df = pd.read_csv(
    "ventas_tienda.csv",
    sep=";",
    skiprows=3,
    na_values=["SIN_DATOS"]
)

print("Resumen de datos")
df.info()

df.drop(["CodigoTienda"], axis = 1, inplace = True)
df.head(7)
```
