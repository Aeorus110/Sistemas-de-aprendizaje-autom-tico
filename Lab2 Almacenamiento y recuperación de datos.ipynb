## Laboratorio 2: Almacenamiento y recuperación de datos en Python

En este ejercicio se mostrará la secuencia de instrucciones necesaria para cargar los datos almacenados en un dataset desde un archivo en formato csv y desde un archivo en formato arff. Veremos como hacerlo cuando los archivos se encuentra en local y utilizando una URL.

Aseguraté de tener instaladas en el sistema las librerias que se importan al inicio del programa y los archivos con los dataset en la misma carpeta que este archivo fuente.



```python
#----------------------------------------------------------------------------
# librerías importadas
import scipy.io.arff as arff
import pandas as pd
import sklearn.datasets as datasets


#------------------------------------------------------------------------------
# Carga de un dataset en local utilizando un archivo .arff
valores, metad = arff.loadarff("./iris.arff") #volcamos los valores y metadatos sobre dos variables
df1 = pd.DataFrame(valores) # Pasamos los datos a un DataFrame
df1['class'] = df1['class'].str.decode("utf-8") # Convierte a string los datos de la columna class
print("\nDatos cargados en df1:")
print(df1.head(15)) #Presentamos las primeras 15 columnas
```

```python
#------------------------------------------------------------------------------
df2 = pd.read_csv("./iris.csv")
print("\nDatos cargados en df2:")
print(df2.head(60))
```

Pregunta: ¿qué código deberíamos añadir a continuación para conseguir eliminar la columna "Id" del df2?


```python
#Añade a continuación el código necesario
df2 = df2.drop(["id"], axis=1)
print(df2.head())
```

```python
#------------------------------------------------------------------------------
# Carga de un datase en formato csv desde una URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv" # URL válida de ejemplo
df3 = pd.read_csv(url)

print("\nDatos cargados en df3 desde una URL donde se localiza el archivo IRIS en formato CSV:")
print(df3.head())
print()
#-------------------------------------------------------------------------------
# De esta forma podemos cargar un dataset desde una librería
data = datasets.load_iris()

# Crear DataFrame con las características (features)
# data.data contiene las columnas con las características predictoras.
# columns=data.feature_names asigna los nombres de las columnas.
df4 = pd.DataFrame(data.data, columns=data.feature_names)

# Añadir la columna target
# en muchos datasets de sklearn, los datos de entrada (features) y la variable objetivo (target)
# vienen separados.
df4['target'] = data.target

print("\nDatos cargados desde sklearn.datasets sobre df4")
print(df4.head())

# Cargamos en df5 solo la coluna "longitud de sepalo"
df5=df4[["sepal length (cm)"]]

# Cargamos en df6 las columnas "longitud del sepalo y ancho del sepalo"
df6=df4[["sepal length (cm)", "sepal width (cm)"]]

print("Datos en df5 que solo contiene la columna sepal length")
print(df5.head())
print()

print("Datos en df6 que solo contiene la columna sepal length")
print(df6.head())
```
